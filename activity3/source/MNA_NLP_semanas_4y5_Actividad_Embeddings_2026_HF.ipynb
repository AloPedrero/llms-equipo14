# **Pregunta - 1:**



Descarga los 3 archivos de la plataforma de la UCI indicado previamente y genera un nuevo DataFrame de Pandas con ellos.

**Llama simplemente "df" a dicho DataFrame.**




```python
PATH = "../files"

df_amazon = pd.read_csv(os.path.join(PATH, 'amazon_cells_labelled.txt'), sep='\t', names=['review','label'], header=None, encoding='utf-8')
dfi_imdb = pd.read_csv(os.path.join(PATH, 'imdb_labelled.txt'), delimiter='\t', names=['review','label'], header=None, encoding='utf-8')
df_yelp = pd.read_csv(os.path.join(PATH, 'yelp_labelled.txt'), sep='\t', names=['review','label'], header=None, encoding='utf-8')
```

```python
newdfi=[]
dfi_list = dfi_imdb.values.tolist()

for row in dfi_list:

    review = str(row[0])
    label = str(row[1])

    text = review + '\t' + label

    matches = re.findall(r'(.*?)\t([01])(?:\n|$)', text, re.DOTALL)

    for rev, lab in matches:

        rev = rev.strip()

        if rev != '':
            newdfi.append([rev, int(lab)])

# convertir a DataFrame
clean_dfi_imdb = pd.DataFrame(newdfi, columns=['review', 'label'])
```

```python
clean_dfi_imdb.info()
```

```python
df = pd.concat([df_amazon, clean_dfi_imdb, df_yelp], ignore_index=True) 
# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 2:**

Proceso de limpieza. Aplica el proceso de limpieza que consideres adecuado.











```python
# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


# ************* Inicia la sección de agregar código:*****************************

X = df.review
y = df.label

assert X.shape == (3000,)
assert y.shape == (3000,)  
```

```python
def clean_tok(doc):
  doc_cleaned = ''.join(
    char for char in doc.lower()
    if char not in string.punctuation and not char.isdigit())
  
  tokens = nltk.word_tokenize(doc_cleaned)
  
  tokens = [
    word for word in tokens
    if len(word) > 1]

  return tokens
```

```python
Xclean = [clean_tok(x) for x in X]

# *********** Aquí termina la sección de agregar código *************
```

```python
# Despleguemos los primeros comentarios después de tu proceso de limpieza:

for x in Xclean[0:5]:
  print(x)
```

```python

# ************* Inicia la sección de agregar código:*****************************

Xtrain, x_val_and_test, Ytrain, y_val_and_test = train_test_split(Xclean, y, train_size=.70, shuffle=True, random_state=1)
Xval, Xtest, Yval, Ytest = train_test_split(x_val_and_test, y_val_and_test, test_size=.50, shuffle=True, random_state=1)
```

# **Pregunta - 4:**




### **Construye tu vocabulario a continuación utilizando solamente el conjunto de Train:**


```python
# a.	Usa el conjunto de entrenamiento para generar tu vocabulario
#     con un tamaño que consideres adecuado:


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********

token_dict = Counter()

for k in range(len(Xtrain)):
  token_dict.update(Xtrain[k])


print('Longitud del diccionario:', len(token_dict))
print('\n(word,frequency):')
print(token_dict.most_common(10))
```

```python

min_freq = 5

midicc = {}

for token, freq in token_dict.items():

    if freq >= min_freq:
        midicc[token] = freq

# *********** Aquí termina la sección de agregar código *************
```

```python
# b.	Indica el tamaño del vocabulario generado.

# ******* Inicia la sección de agregar código: ***********

print('Longitud del vocabulario generado:', len(midicc))

# *********** Aquí termina la sección de agregar código *************
```

```python
# c.	Con el vocabulario generado, filtra los conjuntos de entrenamiento,
#     validación y prueba para que todos los comentarios usen solamente las
#     palabras de este vocabulario.

#     Llamar train_X, val_X y test_X a estos tres conjuntos.


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


train_X = []
for ss in Xtrain:
  train_X.append([w for w in ss if w in midicc])

val_X = []
for ss in Xval:
  val_X.append([w for w in ss if w in midicc])

test_X = []
for ss in Xtest:
  test_X.append([w for w in ss if w in midicc])


# *********** Aquí termina la sección de agregar código *************
```

```python
# Vemos el resultado de los primeros comentarios del conjunto de validación:

for ss in val_X[0:5]:
  print(ss)
```

* **a) bge-base-en-v1.5**
- Es un modelo de embeddings desarrollado por Beijing Academy of Artificial Intelligence (BAAI).
- Soporta tareas de búsqueda semántica (retrieval), similitud textual y sistemas RAG.
- Fue entrenado principalmente para textos en inglés.
- Cuenta con 12 capas (layers).
- El tamaño del vector embebido es de 768 dimensiones.
- Tiene un buen equilibrio entre rendimiento, velocidad y consumo de memoria.


* **b) bge-large-en-v1.5**
- Es una versión más grande y precisa de la familia BGE.
- Está optimizado para tareas de recuperación de información y búsqueda semántica.
- Cuenta con 24 capas (layers).
- El tamaño del embedding es de 1024 dimensiones.
- Genera representaciones semánticas más precisas que la versión base.
- Requiere más memoria RAM/VRAM y mayor tiempo de procesamiento.

* **c)	e5-base-v2**
- Es un modelo de embeddings desarrollado por Microsoft.
- Está diseñado para tareas de recuperación semántica y comparación de textos.
- Cuenta con 12 capas (layers).
El tamaño del vector embebido es de 768 dimensiones.
- Funciona principalmente con textos en inglés.
- Los textos largos se truncan a un máximo de 512 tokens
