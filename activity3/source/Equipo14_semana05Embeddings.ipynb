```python
PATH = 'sentiment labelled sentences'
if not os.path.exists(PATH):
    PATH = '../files'

df_amazon = pd.read_csv(os.path.join(PATH, 'amazon_cells_labelled.txt'), sep='	', names=['review','label'], header=None, encoding='utf-8')
dfi_imdb = pd.read_csv(os.path.join(PATH, 'imdb_labelled.txt'), delimiter='	', names=['review','label'], header=None, encoding='utf-8')
df_yelp = pd.read_csv(os.path.join(PATH, 'yelp_labelled.txt'), sep='	', names=['review','label'], header=None, encoding='utf-8')
```

```python
newdfi=[]
dfi_list = dfi_imdb.values.tolist()

for row in dfi_list:

    review = str(row[0])
    label = str(row[1])

    text = review + '\t' + label

    matches = re.findall(r'(.*?)\t([01])(?:\n|$)', text, re.DOTALL)

    for rev, lab in matches:

        rev = rev.strip()

        if rev != '':
            newdfi.append([rev, int(lab)])

# convertir a DataFrame
clean_dfi_imdb = pd.DataFrame(newdfi, columns=['review', 'label'])
```

# **Pregunta - 2:**

Proceso de limpieza. Aplica el proceso de limpieza que consideres adecuado.











```python
# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


# ************* Inicia la sección de agregar código:*****************************

X = df.review
y = df.label

assert X.shape == (3000,)
assert y.shape == (3000,)  
```

```python
def clean_tok(doc):
  doc_cleaned = ''.join(
    char for char in doc.lower()
    if char not in string.punctuation and not char.isdigit())
  
  tokens = re.findall(r'[a-z]+', doc_cleaned)
  
  tokens = [
    word for word in tokens
    if len(word) > 1]

  return tokens
```

```python
Xclean = [clean_tok(x) for x in X]

# *********** Aquí termina la sección de agregar código *************
```

```python
# Despleguemos los primeros comentarios después de tu proceso de limpieza:

for x in Xclean[0:5]:
  print(x)
```

# **Pregunta - 3:**



Realicemos una partición aleatoria con los porcentajes que consideres más adecuados. Utiliza una semilla para su reproducibilidad.

```python

# ************* Inicia la sección de agregar código:*****************************

Xtrain, x_val_and_test, Ytrain, y_val_and_test = train_test_split(Xclean, y, train_size=.70, shuffle=True, random_state=1)
Xval, Xtest, Yval, Ytest = train_test_split(x_val_and_test, y_val_and_test, test_size=.50, shuffle=True, random_state=1)
```

```python
print('X,y Train:', len(Xtrain), len(Ytrain))
print('X,y Val:', len(Xval), len(Yval))
print('X,y Test', len(Xtest), len(Ytest))

# *********** Termina la sección de agregar código *************
```

# **Pregunta - 4:**




### **Construye tu vocabulario a continuación utilizando solamente el conjunto de Train:**


```python
# a.	Usa el conjunto de entrenamiento para generar tu vocabulario
#     con un tamaño que consideres adecuado:


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********

token_dict = Counter()

for k in range(len(Xtrain)):
  token_dict.update(Xtrain[k])


print('Longitud del diccionario:', len(token_dict))
print('\n(word,frequency):')
print(token_dict.most_common(10))
```

```python

min_freq = 5

midicc = {}

for token, freq in token_dict.items():

    if freq >= min_freq:
        midicc[token] = freq

# *********** Aquí termina la sección de agregar código *************
```

```python
# b.	Indica el tamaño del vocabulario generado.

# ******* Inicia la sección de agregar código: ***********

print('Longitud del vocabulario generado:', len(midicc))

# *********** Aquí termina la sección de agregar código *************
```

```python
# c.	Con el vocabulario generado, filtra los conjuntos de entrenamiento,
#     validación y prueba para que todos los comentarios usen solamente las
#     palabras de este vocabulario.

#     Llamar train_X, val_X y test_X a estos tres conjuntos.


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


train_X = []
for ss in Xtrain:
  train_X.append([w for w in ss if w in midicc])

val_X = []
for ss in Xval:
  val_X.append([w for w in ss if w in midicc])

test_X = []
for ss in Xtest:
  test_X.append([w for w in ss if w in midicc])


# *********** Aquí termina la sección de agregar código *************
```

```python
# Vemos el resultado de los primeros comentarios del conjunto de validación:

for ss in val_X[0:5]:
  print(ss)
```

### ++++++++ Inicia la sección de agregar texto: +++++++++++

* **a) bge-base-en-v1.5**
- Es un modelo de embeddings desarrollado por Beijing Academy of Artificial Intelligence (BAAI).
- Soporta tareas de búsqueda semántica (retrieval), similitud textual y sistemas RAG.
- Fue entrenado principalmente para textos en inglés.
- Cuenta con 12 capas (layers).
- El tamaño del vector embebido es de 768 dimensiones.
- Tiene un buen equilibrio entre rendimiento, velocidad y consumo de memoria.


* **b) bge-large-en-v1.5**
- Es una versión más grande y precisa de la familia BGE.
- Está optimizado para tareas de recuperación de información y búsqueda semántica.
- Cuenta con 24 capas (layers).
- El tamaño del embedding es de 1024 dimensiones.
- Genera representaciones semánticas más precisas que la versión base.
- Requiere más memoria RAM/VRAM y mayor tiempo de procesamiento.

* **c)	e5-base-v2**
- Es un modelo de embeddings desarrollado por Microsoft.
- Está diseñado para tareas de recuperación semántica y comparación de textos.
- Cuenta con 12 capas (layers).
El tamaño del vector embebido es de 768 dimensiones.
- Funciona principalmente con textos en inglés.
- Los textos largos se truncan a un máximo de 512 tokens

# **Pregunta - 7:**




Generamos los vectores embebidos a partir de los conjuntos de entrenamiento, validación y prueba y con las características indicadas en el archivo PDF.

Los llamaremos trainEmb, valEmb y testEmb, respectivamente.


```python
# a) Comentarios con vectores embebidos.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


def doc_a_vector(doc, dic):
    vectores = [dic[w] for w in doc if w in dic]
    if vectores:
        return np.mean(vectores, axis=0)
    return np.zeros(len(next(iter(dic.values()))))

trainEmb = np.array([doc_a_vector(doc, diccionario_emb) for doc in train_X])
valEmb   = np.array([doc_a_vector(doc, diccionario_emb) for doc in val_X])
testEmb  = np.array([doc_a_vector(doc, diccionario_emb) for doc in test_X])


# *********** Aquí termina la sección de agregar código *************
```

```python
# b) Dimensiones de los conjuntos trainEmb, valEmb y testEmb.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


print('trainEmb:', trainEmb.shape)
print('valEmb:  ', valEmb.shape)
print('testEmb: ', testEmb.shape)


# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 8:**

```python
# Número de tokens generedos al obtener cada uno de los conjuntos trainEmb, valEmb y testEmb.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


tokens_train = sum(len([w for w in doc if w in diccionario_emb]) for doc in train_X)
tokens_val   = sum(len([w for w in doc if w in diccionario_emb]) for doc in val_X)
tokens_test  = sum(len([w for w in doc if w in diccionario_emb]) for doc in test_X)

print('Tokens en trainEmb:', tokens_train)
print('Tokens en valEmb:  ', tokens_val)
print('Tokens en testEmb: ', tokens_test)


# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 9:**



Entrenamiento y reporte de los modelos de Regresión Logística y Bosque Aleatorio (Random Forest).


```python
# 9a) REGRESIÓN LOGÍSTICA:

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

modeloLR = LogisticRegression(C=1.0, max_iter=1000, random_state=1)
modeloLR.fit(trainEmb, Ytrain)

acc_train_LR = accuracy_score(Ytrain, modeloLR.predict(trainEmb))
acc_val_LR   = accuracy_score(Yval,   modeloLR.predict(valEmb))

print(f'LR - Exactitud Train: {acc_train_LR:.4f}')
print(f'LR - Exactitud Val:   {acc_val_LR:.4f}')
print()
print(classification_report(Yval, modeloLR.predict(valEmb), target_names=['Negativo', 'Positivo']))


# *********** Aquí termina la sección de agregar código *************
```

```python
# 9b) BOSQUE ALEATORIO (Random Forest):

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


from sklearn.ensemble import RandomForestClassifier

modeloRF = RandomForestClassifier(n_estimators=200, max_depth=5, min_samples_leaf=5, random_state=1)
modeloRF.fit(trainEmb, Ytrain)

acc_train_RF = accuracy_score(Ytrain, modeloRF.predict(trainEmb))
acc_val_RF   = accuracy_score(Yval,   modeloRF.predict(valEmb))

print(f'RF - Exactitud Train: {acc_train_RF:.4f}')
print(f'RF - Exactitud Val:   {acc_val_RF:.4f}')
print()
print(classification_report(Yval, modeloRF.predict(valEmb), target_names=['Negativo', 'Positivo']))


# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 10**

**Proceso basado en modelos Preentrenados**

```python
# 10a) Particion.:

# ******* Incluye a continuacion todas las lineas de codigo y celdas que requieras: ***********


Xtrain_pt, x_val_and_test_pt, Ytrain_pt, y_val_and_test_pt = train_test_split(
    df.review.astype(str),
    df.label,
    train_size=.70,
    shuffle=True,
    random_state=1,
    stratify=df.label
)

Xval_pt, Xtest_pt, Yval_pt, Ytest_pt = train_test_split(
    x_val_and_test_pt,
    y_val_and_test_pt,
    test_size=.50,
    shuffle=True,
    random_state=1,
    stratify=y_val_and_test_pt
)

print('X,y Train:', len(Xtrain_pt), len(Ytrain_pt))
print('X,y Val:  ', len(Xval_pt), len(Yval_pt))
print('X,y Test: ', len(Xtest_pt), len(Ytest_pt))
print('Distribucion Train:', Ytrain_pt.value_counts().sort_index().to_dict())
print('Distribucion Val:  ', Yval_pt.value_counts().sort_index().to_dict())
print('Distribucion Test: ', Ytest_pt.value_counts().sort_index().to_dict())


# *********** Aqui termina la seccion de agregar codigo *************
```

```python
# 10b) Vectores embebidos:

# ******* Incluye a continuacion todas las lineas de codigo y celdas que requieras: ***********


# En este enfoque se usa directamente cada resena completa como entrada del
# modelo preentrenado, sin promediar embeddings de palabras del vocabulario.
modelo_emb_pt = modelo_emb if 'modelo_emb' in globals() else SentenceTransformer('BAAI/bge-base-en-v1.5')

trainEmb_pt = modelo_emb_pt.encode(
    Xtrain_pt.tolist(), batch_size=32, show_progress_bar=True, normalize_embeddings=True
)
valEmb_pt = modelo_emb_pt.encode(
    Xval_pt.tolist(), batch_size=32, show_progress_bar=True, normalize_embeddings=True
)
testEmb_pt = modelo_emb_pt.encode(
    Xtest_pt.tolist(), batch_size=32, show_progress_bar=True, normalize_embeddings=True
)

print('trainEmb_pt:', trainEmb_pt.shape)
print('valEmb_pt:  ', valEmb_pt.shape)
print('testEmb_pt: ', testEmb_pt.shape)


# *********** Aqui termina la seccion de agregar codigo *************
```

```python
# 10c) REGRESION LOGISTICA.

# ******* Incluye a continuacion todas las lineas de codigo y celdas que requieras: ***********


modeloLR_pt = LogisticRegression(C=1.0, max_iter=1000, random_state=1)
modeloLR_pt.fit(trainEmb_pt, Ytrain_pt)

acc_train_LR_pt = accuracy_score(Ytrain_pt, modeloLR_pt.predict(trainEmb_pt))
acc_val_LR_pt = accuracy_score(Yval_pt, modeloLR_pt.predict(valEmb_pt))

print(f'LR PT - Exactitud Train: {acc_train_LR_pt:.4f}')
print(f'LR PT - Exactitud Val:   {acc_val_LR_pt:.4f}')
print()
print(classification_report(Yval_pt, modeloLR_pt.predict(valEmb_pt), target_names=['Negativo', 'Positivo']))


# *********** Aqui termina la seccion de agregar codigo *************
```

```python
# 10d) BOSQUE ALEATORIO.

# ******* Incluye a continuacion todas las lineas de codigo y celdas que requieras: ***********


modeloRF_pt = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=3,
    random_state=1,
    n_jobs=-1
)
modeloRF_pt.fit(trainEmb_pt, Ytrain_pt)

acc_train_RF_pt = accuracy_score(Ytrain_pt, modeloRF_pt.predict(trainEmb_pt))
acc_val_RF_pt = accuracy_score(Yval_pt, modeloRF_pt.predict(valEmb_pt))

print(f'RF PT - Exactitud Train: {acc_train_RF_pt:.4f}')
print(f'RF PT - Exactitud Val:   {acc_val_RF_pt:.4f}')
print()
print(classification_report(Yval_pt, modeloRF_pt.predict(valEmb_pt), target_names=['Negativo', 'Positivo']))


# *********** Aqui termina la seccion de agregar codigo *************
```

```python
# Reporte del mejor modelo y particion con el conjunto de Prueba.

# ******* Incluye a continuacion todas las lineas de codigo y celdas que requieras: ***********


resultados_validacion = pd.DataFrame([
    {'Representacion': 'Promedio de embeddings por vocabulario', 'Modelo': 'Regresion Logistica', 'Train': acc_train_LR, 'Validacion': acc_val_LR},
    {'Representacion': 'Promedio de embeddings por vocabulario', 'Modelo': 'Bosque Aleatorio', 'Train': acc_train_RF, 'Validacion': acc_val_RF},
    {'Representacion': 'Embeddings preentrenados directos', 'Modelo': 'Regresion Logistica', 'Train': acc_train_LR_pt, 'Validacion': acc_val_LR_pt},
    {'Representacion': 'Embeddings preentrenados directos', 'Modelo': 'Bosque Aleatorio', 'Train': acc_train_RF_pt, 'Validacion': acc_val_RF_pt},
])

print('Comparacion en validacion:')
print(resultados_validacion.sort_values('Validacion', ascending=False).to_string(index=False))

mejor_fila = resultados_validacion.sort_values('Validacion', ascending=False).iloc[0]
print()
print('Mejor modelo seleccionado por validacion:')
print(f"{mejor_fila['Modelo']} con {mejor_fila['Representacion']}")

if mejor_fila['Representacion'] == 'Embeddings preentrenados directos':
    mejor_modelo = modeloLR_pt if mejor_fila['Modelo'] == 'Regresion Logistica' else modeloRF_pt
    mejor_testEmb = testEmb_pt
    mejor_Ytest = Ytest_pt
else:
    mejor_modelo = modeloLR if mejor_fila['Modelo'] == 'Regresion Logistica' else modeloRF
    mejor_testEmb = testEmb
    mejor_Ytest = Ytest

y_pred_test_mejor = mejor_modelo.predict(mejor_testEmb)
acc_test_mejor = accuracy_score(mejor_Ytest, y_pred_test_mejor)

print()
print(f'Exactitud del mejor modelo en Prueba: {acc_test_mejor:.4f}')
print()
print(classification_report(mejor_Ytest, y_pred_test_mejor, target_names=['Negativo', 'Positivo']))


# *********** Aqui termina la seccion de agregar codigo *************
```

- La primera particion fue **70% entrenamiento / 15% validacion / 15% prueba** con `random_state=1`, resultando en 2100, 450 y 450 muestras respectivamente.

### Enfoques de embeddings evaluados

- En las preguntas 6 a 9 se genero un diccionario `{palabra: vector}` con **bge-base-en-v1.5** y cada resena se represento como el promedio de los embeddings de sus tokens presentes en el vocabulario.
- En la pregunta 10 se uso un segundo enfoque basado directamente en modelos preentrenados: cada resena completa se envio a **bge-base-en-v1.5** y se obtuvo un embedding semantico de 768 dimensiones para el documento completo.
- Para el enfoque preentrenado directo se utilizo una particion estratificada 70/15/15, manteniendo 1050 resenas negativas y 1050 positivas en entrenamiento, y 225/225 en validacion y prueba.

### Resultados principales

| Representacion | Modelo | Exactitud Train | Exactitud Val |
|---|---:|---:|---:|
| Promedio de embeddings por vocabulario | Regresion Logistica | 80.81% | 76.89% |

- Al evaluarlo en el conjunto de prueba, el mejor modelo obtuvo **96.89%** de exactitud, con precision, recall y F1 cercanos a **0.97** en ambas clases.
- Random Forest tambien tuvo muy buen desempeno con embeddings directos, pero mostro una brecha train-validacion mayor que Regresion Logistica, por lo que la Regresion Logistica ofrece una solucion mas simple y con mejor generalizacion.

### Conclusion

Los embeddings preentrenados de Hugging Face permiten representar mejor el significado de comentarios cortos que una agregacion promedio de vectores por palabra. Para este conjunto de datos balanceado, una Regresion Logistica entrenada sobre embeddings directos de **bge-base-en-v1.5** fue suficiente para obtener un desempeno alto, estable y equilibrado entre clases positivas y negativas.
