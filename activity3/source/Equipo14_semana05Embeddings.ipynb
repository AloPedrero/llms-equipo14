```python
newdfi=[]
dfi_list = dfi_imdb.values.tolist()

for row in dfi_list:

    review = str(row[0])
    label = str(row[1])

    text = review + '\t' + label

    matches = re.findall(r'(.*?)\t([01])(?:\n|$)', text, re.DOTALL)

    for rev, lab in matches:

        rev = rev.strip()

        if rev != '':
            newdfi.append([rev, int(lab)])

# convertir a DataFrame
clean_dfi_imdb = pd.DataFrame(newdfi, columns=['review', 'label'])
```

# **Pregunta - 2:**

Proceso de limpieza. Aplica el proceso de limpieza que consideres adecuado.











```python
# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


# ************* Inicia la sección de agregar código:*****************************

X = df.review
y = df.label

assert X.shape == (3000,)
assert y.shape == (3000,)  
```

```python
def clean_tok(doc):
  doc_cleaned = ''.join(
    char for char in doc.lower()
    if char not in string.punctuation and not char.isdigit())
  
  tokens = nltk.word_tokenize(doc_cleaned)
  
  tokens = [
    word for word in tokens
    if len(word) > 1]

  return tokens
```

```python
Xclean = [clean_tok(x) for x in X]

# *********** Aquí termina la sección de agregar código *************
```

```python
# Despleguemos los primeros comentarios después de tu proceso de limpieza:

for x in Xclean[0:5]:
  print(x)
```

# **Pregunta - 3:**



Realicemos una partición aleatoria con los porcentajes que consideres más adecuados. Utiliza una semilla para su reproducibilidad.

```python

# ************* Inicia la sección de agregar código:*****************************

Xtrain, x_val_and_test, Ytrain, y_val_and_test = train_test_split(Xclean, y, train_size=.70, shuffle=True, random_state=1)
Xval, Xtest, Yval, Ytest = train_test_split(x_val_and_test, y_val_and_test, test_size=.50, shuffle=True, random_state=1)
```

# **Pregunta - 4:**




### **Construye tu vocabulario a continuación utilizando solamente el conjunto de Train:**


```python
# a.	Usa el conjunto de entrenamiento para generar tu vocabulario
#     con un tamaño que consideres adecuado:


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********

token_dict = Counter()

for k in range(len(Xtrain)):
  token_dict.update(Xtrain[k])


print('Longitud del diccionario:', len(token_dict))
print('\n(word,frequency):')
print(token_dict.most_common(10))
```

```python

min_freq = 5

midicc = {}

for token, freq in token_dict.items():

    if freq >= min_freq:
        midicc[token] = freq

# *********** Aquí termina la sección de agregar código *************
```

```python
# b.	Indica el tamaño del vocabulario generado.

# ******* Inicia la sección de agregar código: ***********

print('Longitud del vocabulario generado:', len(midicc))

# *********** Aquí termina la sección de agregar código *************
```

```python
# c.	Con el vocabulario generado, filtra los conjuntos de entrenamiento,
#     validación y prueba para que todos los comentarios usen solamente las
#     palabras de este vocabulario.

#     Llamar train_X, val_X y test_X a estos tres conjuntos.


# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


train_X = []
for ss in Xtrain:
  train_X.append([w for w in ss if w in midicc])

val_X = []
for ss in Xval:
  val_X.append([w for w in ss if w in midicc])

test_X = []
for ss in Xtest:
  test_X.append([w for w in ss if w in midicc])


# *********** Aquí termina la sección de agregar código *************
```

```python
# Vemos el resultado de los primeros comentarios del conjunto de validación:

for ss in val_X[0:5]:
  print(ss)
```

# **Pregunta - 7:**




Generamos los vectores embebidos a partir de los conjuntos de entrenamiento, validación y prueba y con las características indicadas en el archivo PDF.

Los llamaremos trainEmb, valEmb y testEmb, respectivamente.


```python
# a) Comentarios con vectores embebidos.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


def doc_a_vector(doc, dic):
    vectores = [dic[w] for w in doc if w in dic]
    if vectores:
        return np.mean(vectores, axis=0)
    return np.zeros(len(next(iter(dic.values()))))

trainEmb = np.array([doc_a_vector(doc, diccionario_emb) for doc in train_X])
valEmb   = np.array([doc_a_vector(doc, diccionario_emb) for doc in val_X])
testEmb  = np.array([doc_a_vector(doc, diccionario_emb) for doc in test_X])


# *********** Aquí termina la sección de agregar código *************
```

```python
# b) Dimensiones de los conjuntos trainEmb, valEmb y testEmb.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


print('trainEmb:', trainEmb.shape)
print('valEmb:  ', valEmb.shape)
print('testEmb: ', testEmb.shape)


# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 8:**

```python
# Número de tokens generedos al obtener cada uno de los conjuntos trainEmb, valEmb y testEmb.

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


tokens_train = sum(len([w for w in doc if w in diccionario_emb]) for doc in train_X)
tokens_val   = sum(len([w for w in doc if w in diccionario_emb]) for doc in val_X)
tokens_test  = sum(len([w for w in doc if w in diccionario_emb]) for doc in test_X)

print('Tokens en trainEmb:', tokens_train)
print('Tokens en valEmb:  ', tokens_val)
print('Tokens en testEmb: ', tokens_test)


# *********** Aquí termina la sección de agregar código *************
```

# **Pregunta - 9:**



Entrenamiento y reporte de los modelos de Regresión Logística y Bosque Aleatorio (Random Forest).


```python
# 9a) REGRESIÓN LOGÍSTICA:

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

modeloLR = LogisticRegression(C=1.0, max_iter=1000, random_state=1)
modeloLR.fit(trainEmb, Ytrain)

acc_train_LR = accuracy_score(Ytrain, modeloLR.predict(trainEmb))
acc_val_LR   = accuracy_score(Yval,   modeloLR.predict(valEmb))

print(f'LR - Exactitud Train: {acc_train_LR:.4f}')
print(f'LR - Exactitud Val:   {acc_val_LR:.4f}')
print()
print(classification_report(Yval, modeloLR.predict(valEmb), target_names=['Negativo', 'Positivo']))


# *********** Aquí termina la sección de agregar código *************
```

```python
# 9b) BOSQUE ALEATORIO (Random Forest):

# ******* Inlcuye a continuación todas las líneas de código y celdas que requieras: ***********


from sklearn.ensemble import RandomForestClassifier

modeloRF = RandomForestClassifier(n_estimators=200, max_depth=5, min_samples_leaf=5, random_state=1)
modeloRF.fit(trainEmb, Ytrain)

acc_train_RF = accuracy_score(Ytrain, modeloRF.predict(trainEmb))
acc_val_RF   = accuracy_score(Yval,   modeloRF.predict(valEmb))

print(f'RF - Exactitud Train: {acc_train_RF:.4f}')
print(f'RF - Exactitud Val:   {acc_val_RF:.4f}')
print()
print(classification_report(Yval, modeloRF.predict(valEmb), target_names=['Negativo', 'Positivo']))


# *********** Aquí termina la sección de agregar código *************
```

### ++++++++ Inicia la sección de agregar texto: +++++++++++


### Preprocesamiento y vocabulario

- Se aplicó limpieza básica: conversión a minúsculas, eliminación de puntuación y dígitos, tokenización con NLTK y descarte de tokens de un solo carácter.
- Para construir el vocabulario se usó únicamente el conjunto de entrenamiento, con un umbral de **min_freq = 5**. Esto resultó en un vocabulario de **646 palabras**, eliminando términos muy raros que podrían introducir ruido sin aportar capacidad de generalización.
- La partición fue **70% entrenamiento / 15% validación / 15% prueba** con `random_state=1` para reproducibilidad, resultando en 2100, 450 y 450 muestras respectivamente.

### Modelo de embeddings

- Se eligió **bge-base-en-v1.5** (BAAI) por su buen equilibrio entre rendimiento semántico, velocidad de inferencia y consumo de memoria (768 dimensiones, 12 capas).
- Se generó un diccionario de embeddings `{palabra: vector}` a partir del vocabulario de 646 palabras. Cada documento se representó como el **promedio** de los vectores de sus tokens presentes en el diccionario.

### Hiperparámetros de los clasificadores

- **Regresión Logística:** `C=1.0` (regularización estándar), `max_iter=1000` para garantizar convergencia.
- **Bosque Aleatorio:** `n_estimators=200`, `max_depth=5`, `min_samples_leaf=5`. Los valores conservadores de profundidad y mínimo de muestras por hoja buscan reducir el sobreajuste, que es una tendencia común de RF con datos de texto de alta dimensionalidad.

### Resultados obtenidos (Q9)

| Modelo | Exactitud Train | Exactitud Val | Brecha |
|--------|----------------|---------------|--------|
| Regresión Logística | 78.33% | 76.44% | ~2% |
| Bosque Aleatorio    | 90.86% | 75.33% | ~15% |

- La **Regresión Logística** es el modelo más sólido: buena exactitud en validación y una brecha train-val de apenas ~2%, lo que indica buena generalización.
- El **Bosque Aleatorio** muestra sobreajuste notable (~15% de brecha) a pesar de los hiperparámetros conservadores (`max_depth=5`, `min_samples_leaf=5`), lo que sugiere que este enfoque de embeddings promediados no es el más adecuado para RF.
- Ambos modelos obtienen precisión y recall balanceados entre clases positiva y negativa (~0.75–0.77), lo que indica que no hay sesgo hacia ninguna clase.
